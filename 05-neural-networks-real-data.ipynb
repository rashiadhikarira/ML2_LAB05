## 3. Neural Network (40 points)

Implement a neural network as you did in the previous lab, but this time leverage the new functionalities introduced this week. Specifically:

- Your class implementing the neural network should inherit from nn.Module.

- Use `nn.Linear` for the linear transformations between layers, `nn.ReLU` for the hidden-layer activations, and the appropriate activation function for the output layer (e.g., `nn.Sigmoid` for binary classification).

- Use the appropriate cost function from `torch.nn` for your task (e.g., `nn.MSELoss`, `nn.BCELoss`).

Once you have implemented this class, create an instance of it, then train and evaluate your model on the lab dataset.

**Remember to follow the general requirements regarding model tuning, evaluation, and data preprocessing described earlier.**

```python
# USE AS MANY CELLS AS NEEDED
import torch
from torch import nn, optim
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from torch.utils.data import TensorDataset, DataLoader
from sklearn.preprocessing import StandardScaler

class MLPRegression(nn.Module):
    def __init__(self, input_dimensions, lr=0.01, epochs=150):
        super().__init__()

        self.lr, self.epochs = lr, epochs

        self.relu = nn.ReLU()
        self.linear_1 = nn.Linear(input_dimensions, 100) 
        self.linear_2 = nn.Linear(100, 100) 
        self.linear_3 = nn.Linear(100, 10) 
        self.linear_4 = nn.Linear(10, 1) 

        self.loss_fn = nn.MSELoss()
        self.optimizer = torch.optim.Adam(self.parameters(), lr=self.lr)

    def forward(self, x):
        x = self.linear_1(x)
        x = self.relu(x)
        x = self.linear_2(x)
        x = self.relu(x)
        x = self.linear_3(x)
        x = self.relu(x)
        x = self.linear_4(x)
        return x

    def fit(self, dataset, loader):
        history = {"train": []}

        for epoch in range(self.epochs):
            self.train()
            total = 0.0
            for xb, yb in loader:
                self.optimizer.zero_grad()
                loss = self.cost(self(xb), yb)
                loss.backward()
                self.optimizer.step()
                total += loss.item() * len(xb)
            history["train"].append(total / len(dataset))
                    
        return history

    def cost(self, y_pred, y_true):
        return self.loss_fn(y_pred, y_true)

    def predict(self, X):
        self.eval()
        with torch.no_grad():
            return self(X) # this calls forward

X = df.drop(columns='Concrete compressive strength(MPa, megapascals)')
y = df['Concrete compressive strength(MPa, megapascals)']

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=42) 

batch_size = 64

scaler = StandardScaler()
X_train_changed = scaler.fit_transform(X_train)
X_test_changed = scaler.transform(X_test)

X_train_changed = torch.tensor(X_train_changed, dtype = torch.float32)
X_test_changed = torch.tensor(X_test_changed, dtype = torch.float32)
y_train = torch.tensor(y_train.values, dtype = torch.float32).view(-1, 1)
y_test =  torch.tensor(y_test.values, dtype = torch.float32).view(-1, 1)

train_data = TensorDataset(X_train_changed, y_train)
train_loader = DataLoader(train_data, batch_size=batch_size, shuffle=True)

test_data = TensorDataset(X_test_changed, y_test)
test_loader = DataLoader(test_data, batch_size=batch_size, shuffle=False)

model = MLPRegression(X_train_changed.shape[1], lr = 1e-3, epochs = 150)
history = model.fit(train_data, train_loader)
y_pred = model.predict(X_test_changed)
cost = mean_squared_error(y_test.numpy(), y_pred.numpy())

print(f"The Regression models total cost is {cost}")

plt.plot(history["train"])
plt.xlabel("epoch")
plt.ylabel("train MSE")
plt.show()
```

## 4. Replication (10 points)

Replicate your neural network experiment using the corresponding model class from `scikit-learn`: https://scikit-learn.org/stable/api/sklearn.neural_network.html. Be sure to apply the same neural network architecture as well as the same procedures for model evaluation and preprocessing to ensure a fair comparison.

**Remember**: Follow the general requirements regarding model tuning, evaluation, and data preprocessing described earlier.

```python
import warnings
import numpy as np
from sklearn.model_selection import KFold
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_squared_error
from sklearn.exceptions import ConvergenceWarning

warnings.filterwarnings("ignore", category=ConvergenceWarning)

X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

kf = KFold(n_splits=5, shuffle=True, random_state=42)
sk_mse = []

for fold, (train_idx, test_idx) in enumerate(kf.split(X), 1):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    # scale inside each fold, fit on the training part only
    scaler = StandardScaler()
    X_train = scaler.fit_transform(X_train)
    X_test = scaler.transform(X_test)

    # same architecture as the PyTorch model: 100 -> 100 -> 10 -> 1, ReLU, Adam
    sk_model = MLPRegressor(
        hidden_layer_sizes=(100, 100, 10),
        activation="relu",
        solver="adam",
        learning_rate_init=0.01,
        batch_size=32,
        max_iter=150,
        alpha=0.0,              # no L2 penalty (the PyTorch model has none)
        n_iter_no_change=150,   # no early stopping, train all 150 epochs
        random_state=42,
    )
    sk_model.fit(X_train, y_train)

    mse = mean_squared_error(y_test, sk_model.predict(X_test))
    sk_mse.append(mse)
    print(f"Fold {fold}: MSE = {mse:.4f}")

print(f"\nscikit-learn MLPRegressor, average MSE across 5 folds = {np.mean(sk_mse):.4f}")
```

### Replication Results

The scikit-learn MLPRegressor used the same architecture as the PyTorch network (hidden layers of 100, 100, and 10 units with ReLU, one linear output unit, and the Adam optimizer with a learning rate of 0.01, batch size 32, and 150 epochs). It was evaluated with the same 5-fold cross-validation, with scaling done inside each fold to avoid leakage. Its average MSE was 25.33, compared to 26.31 for the PyTorch network, so the two implementations performed very similarly. The small gap is expected, since the two libraries initialize weights differently and shuffle batches in a different order, but the model and the training procedure are otherwise equivalent.

## 5. Comparison (20 points)

Apply a Support Vector Machine (SVM) and Linear Regression model to the dataset, and compare their performance with that of your Neural Network.

Discuss any similarities or differences you observe in their results and provide possible explanations for these patterns.

Be sure to apply the k-fold cross-validation procedure to ensure a fair comparison across all models.

**Remember**: Follow the general requirements regarding model tuning, evaluation, and data preprocessing described earlier.

```python

from sklearn.model_selection import KFold
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from torch.utils.data import TensorDataset, DataLoader

X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

kf = KFold(n_splits=5, shuffle=True, random_state=42)

results = {"Neural Network": [], "SVM": [], "Linear Regression": []}

for train_idx, test_idx in kf.split(X):
    X_train, X_test = X[train_idx], X[test_idx]
    y_train, y_test = y[train_idx], y[test_idx]

    scaler = StandardScaler()
    X_train = scaler.fit_transform(X_train)
    X_test = scaler.transform(X_test)

    # Neural Network
    X_train_t = torch.tensor(X_train, dtype=torch.float32)
    X_test_t = torch.tensor(X_test, dtype=torch.float32)
    y_train_t = torch.tensor(y_train, dtype=torch.float32).reshape(-1, 1)

    dataset = TensorDataset(X_train_t, y_train_t)
    loader = DataLoader(dataset, batch_size=32, shuffle=True)

    torch.manual_seed(42)
    nn_model = MLPRegression(input_dimensions=X.shape[1])
    nn_model.fit(dataset, loader)

    nn_pred = nn_model.predict(X_test_t).numpy().flatten()
    results["Neural Network"].append(mean_squared_error(y_test, nn_pred))

    # Support Vector Machine
    svm = SVR(kernel="rbf")
    svm.fit(X_train, y_train)
    svm_pred = svm.predict(X_test)
    results["SVM"].append(mean_squared_error(y_test, svm_pred))

    # Linear Regression
    lr = LinearRegression()
    lr.fit(X_train, y_train)
    lr_pred = lr.predict(X_test)
    results["Linear Regression"].append(mean_squared_error(y_test, lr_pred))

# Average MSE across all 5 folds
for model, scores in results.items():
    print(f"{model}: Average MSE = {np.mean(scores):.4f}")
```

### Model Performance Comparison 

Using 5-fold cross-validation, the Neural Network had the lowest average MSE (26.3147), followed by SVM (98.6902) and Linear Regression (111.0109). The lower MSE indicates better predictive accuracy; the Neural Network performed best, while Linear Regression performed worst. One possible explanation is that the compressive strength depends on complex and nonlinear relationships between different materials and age. The Neural Network can capture these relationships more effectively, while Linear Regression assumes only linear relationships between the input variables and compressive strength. SVM can also capture nonlinear relationships, which can explain why it performed better than linear regression. However, it still had a higher MSE than the Neural Network. Overall, the results show that the Neural Network was better at predicting the concrete compressive strength compared to the other two models.
